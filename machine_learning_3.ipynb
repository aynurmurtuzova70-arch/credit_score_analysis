```python
import pandas as pd
```

```python
df = pd.read_csv('train.csv')
df
```

```python
pd.set_option('display.max_column', None)
```

```python
df.drop(['ID', 'Customer_ID', 'Name', 'SSN', 'Type_of_Loan', 'Credit_Mix'], axis=1, inplace=True)
```

```python
df['Credit_History_Age']=df['Credit_History_Age'].apply(lambda x: str(x).split(' and '))
```

```python
df['Credit_History_Age'].apply(lambda x: x[-1])
```

```python
df['years']=pd.to_numeric(df['Credit_History_Age'].apply(lambda x: x[0]).str.replace('Years', ''), errors='coerce')
df['months']=pd.to_numeric(df['Credit_History_Age'].apply(lambda x: x[-1]).str.replace('Months', ''), errors='coerce')
```

```python
df.drop(['Credit_History_Age'], axis=1, inplace=True)
```

```python
df.head(1)
```

```python
cols_to_change=['Age', 'Annual_Income', 'Num_of_Loan', 'Num_of_Delayed_Payment', 'Changed_Credit_Limit', 'Outstanding_Debt', 'Amount_invested_monthly' ]
```

```python
for col in cols_to_change:
  df[col] = pd.to_numeric(df[col].str.replace('__-333333333333333333333333333__', '').str.replace('_', ''))
```
